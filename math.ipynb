```python
# 1. Area of Rectangle
length = float(input("Enter length: "))
width = float(input("Enter width: "))
area = length * width
print("Area =", area)


# 2. Circumference of Circle
import math
r = float(input("Enter radius: "))
circumference = 2 * math.pi * r
print("Circumference =", circumference)


# 3. Simple Interest
principal = float(input("Enter Principal: "))
rate = float(input("Enter Rate: "))
time = float(input("Enter Time: "))
simple_interest = principal * rate * time
print("Simple Interest =", simple_interest)


# 4. Speed
distance = float(input("Enter Distance: "))
time = float(input("Enter Time: "))
speed = distance / time
print("Speed =", speed)


# 5. BMI
weight = float(input("Enter Weight: "))
height = float(input("Enter Height: "))
bmi = weight / (height ** 2)
print("BMI =", bmi)


# 6. Force
m = float(input("Enter Mass: "))
a = float(input("Enter Acceleration: "))
force = m * a
print("Force =", force)


# 7. Compound Interest
P = float(input("Enter Principal: "))
r = float(input("Enter Rate: "))
n = int(input("Enter Compounding times: "))
t = float(input("Enter Time: "))
A = P * (1 + r / n) ** (n * t)
print("Total Amount =", A)


# 8. Perimeter of Triangle
a = float(input("Enter side a: "))
b = float(input("Enter side b: "))
c = float(input("Enter side c: "))
perimeter = a + b + c
print("Perimeter =", perimeter)


# 9. Volume of Sphere
r = float(input("Enter radius: "))
volume = (4 / 3) * math.pi * (r ** 3)
print("Volume =", volume)


# 10. Kinetic Energy
m = float(input("Enter Mass: "))
v = float(input("Enter Velocity: "))
KE = 0.5 * m * (v ** 2)
print("Kinetic Energy =", KE)


# 11. Quadratic Equation
a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))
d = b ** 2 - 4 * a * c
x1 = (-b + math.sqrt(d)) / (2 * a)
x2 = (-b - math.sqrt(d)) / (2 * a)
print("Root 1 =", x1)
print("Root 2 =", x2)


# 12. Celsius to Fahrenheit
C = float(input("Enter Celsius: "))
F = (C * 9 / 5) + 32
print("Fahrenheit =", F)


# 13. Gravitational Force
G = 6.67430e-11
m1 = float(input("Enter m1: "))
m2 = float(input("Enter m2: "))
r = float(input("Enter distance: "))
F = G * m1 * m2 / (r ** 2)
print("Gravitational Force =", F)


# 14. Volume of Cylinder
r = float(input("Enter radius: "))
h = float(input("Enter height: "))
volume = math.pi * (r ** 2) * h
print("Volume =", volume)


# 15. Pressure
F = float(input("Enter Force: "))
A = float(input("Enter Area: "))
P = F / A
print("Pressure =", P)


# 16. Electric Power
V = float(input("Enter Voltage: "))
I = float(input("Enter Current: "))
P = V * I
print("Electric Power =", P)


# 17. Perimeter of Circle
r = float(input("Enter radius: "))
perimeter = 2 * math.pi * r
print("Perimeter =", perimeter)


# 18. Future Value
PV = float(input("Enter Present Value: "))
r = float(input("Enter Rate: "))
t = float(input("Enter Time: "))
FV = PV * (1 + r) ** t
print("Future Value =", FV)


# 19. Work Done
f = float(input("Enter Force: "))
d = float(input("Enter Distance: "))
theta = float(input("Enter Angle: "))
W = f * d * math.cos(math.radians(theta))
print("Work Done =", W)


# 20. Heat Transfer
m = float(input("Enter Mass: "))
c = float(input("Enter Specific Heat: "))
T = float(input("Enter Change in Temperature: "))
Q = m * c * T
print("Heat Transfer =", Q)
```
